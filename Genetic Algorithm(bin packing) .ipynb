```python
import random

items = [4, 8, 1, 4, 2, 1]
capacity = 10
population_size = 10
generations = 50


def create_solution():
    bins = []

    for item in items:
        placed = False

        for b in bins:
            if sum(b) + item <= capacity:
                b.append(item)
                placed = True
                break

        if not placed:
            bins.append([item])

    return bins


def fitness(solution):
    return len(solution)


population = [create_solution() for _ in range(population_size)]

for generation in range(generations):
    population.sort(key=fitness)

    best = population[0]

    new_population = [best]

    while len(new_population) < population_size:
        child = create_solution()
        new_population.append(child)

    population = new_population

best_solution = min(population, key=fitness)

print("Best Packing:")
for i, b in enumerate(best_solution, 1):
    print("Bin", i, ":", b, "Total =", sum(b))

print("Number of bins:", len(best_solution))
```
